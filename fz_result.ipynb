```python
import pandas as pd

file_path = "test_results_with_predictions.csv"
df = pd.read_csv(file_path)

def compute_accuracy_and_flags(df, answer_col, pred_col):
    correct_flags = []
    for ans, pred in zip(df[answer_col], df[pred_col]):
        if pd.isna(pred) or pd.isna(ans):
            correct_flags.append(False)
            continue

        ans_norm = str(ans).strip().lower()
        pred_norm = str(pred).strip().lower()

        match = ans_norm in pred_norm
        correct_flags.append(match)

    accuracy = sum(correct_flags) / len(correct_flags)
    return accuracy, correct_flags

acc_zero, flags_zero = compute_accuracy_and_flags(df, "answer", "zeroshot_pred")
acc_few, flags_few = compute_accuracy_and_flags(df, "answer", "fewshot_pred")

df["zeroshot_correct"] = flags_zero
df["fewshot_correct"] = flags_few
print(f"Zeroshot Accuracy: {acc_zero:.2f}")
print(f"Fewshot Accuracy: {acc_few:.2f}")
```
